```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
BP_OG = pd.read_excel("data_for_EEG/BP_OG.xlsx",sheet_name="Горизонтальная")
BP_ZG = pd.read_excel("data_for_EEG/BP_ZG.xlsx",sheet_name="Горизонтальная")
control_OG = pd.read_excel("data_for_EEG/Kontrol_OG.xlsx",sheet_name="Горизонтальная")
control_ZG = pd.read_excel("data_for_EEG/Kontorol_ZG.xlsx",sheet_name="Горизонтальная")
```

```python
import pandas as pd
import os

# Список ваших файлов (только страницы "Горизонтальная")
BP_OG["здоров(1)/болен(0)"] = 0
BP_ZG["здоров(1)/болен(0)"] = 0
control_OG["здоров(1)/болен(0)"] = 1
control_ZG["здоров(1)/болен(0)"] = 1
files = [
BP_OG,
BP_ZG,
control_OG,
control_ZG
]

all_dfs = []

for file in files:
    # Загружаем CSV
    # df = pd.read_csv(file)
    df = file
        
    all_dfs.append(df)

# Объединяем все таблицы в одну
# Используем ignore_index=True, чтобы пересобрать индексы строк
final_df = pd.concat(all_dfs, ignore_index=True, sort=False)

# Сохраняем результат в Excel
final_df = final_df.drop("Имя",axis = 1)
final_df.to_csv("combined_horizontal_data.csv", index=False)

print("Файлы успешно объединены в 'combined_horizontal_data.xlsx'")
```

```python
combined_horizontal_data = pd.read_csv("data_for_EEG/combined_horizontal_data.csv")
combined_horizontal_data
```

```python
import pandas as pd

# Разделяем на два класса
df_0 = combined_horizontal_data[combined_horizontal_data['здоров(1)/болен(0)'] == 0]
df_1 = combined_horizontal_data[combined_horizontal_data['здоров(1)/болен(0)'] == 1]

# Случайно выбираем из нулевого класса столько же строк, сколько в первом
df_0_downsampled = df_0.sample(n=len(df_1), random_state=42)

# Соединяем обратно
df_balanced = pd.concat([df_0_downsampled, df_1])

# Перемешиваем строки
df_balanced = df_balanced.sample(frac=1, random_state=42).reset_index(drop=True)
df_balanced.to_csv("df_balanced.csv", index=False)
df_balanced
```

```python
df_0_remaining = df_0.drop(df_0_downsampled.index)

# Сохраняем, если нужно
df_0_remaining.to_csv("data_for_EEG/df_leftovers.csv", index=False)
```
