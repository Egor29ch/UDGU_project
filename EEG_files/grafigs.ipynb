```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
BP_OG = pd.read_excel("data_for_EEG\BP_OG.xlsx",sheet_name="Вертикальная")
BP_ZG = pd.read_excel("data_for_EEG\BP_ZG.xlsx",sheet_name="Вертикальная")
# axis=1 означает соединение по горизонтали (колонки к колонкам)
combined_df = pd.concat([BP_ZG, BP_OG], axis=1)

# Если появились дубликаты колонок, их можно удалить так:
combined_df = combined_df.loc[:, ~combined_df.columns.duplicated()]
combined_df
combined_df = combined_df.drop("Имя",axis=1)

pearson_corr_BP_OG = combined_df.corr(method='pearson')
print("Корреляция Пирсона:")
print(pearson_corr_BP_OG)
plt.figure(figsize=(20, 20))

# Создаем тепловую карту
sns.heatmap(pearson_corr_BP_OG, annot=True, cmap='coolwarm', vmin=-1, vmax=1, fmt=".2f")
plt.title('Корреляционная матрица признаков датасета')
plt.tight_layout()
plt.show()
```

```python
С_OG = pd.read_excel("data_for_EEG\Kontorol_ZG.xlsx",sheet_name="Вертикальная")
С_ZG = pd.read_excel("data_for_EEG\Kontrol_OG.xlsx",sheet_name="Вертикальная")
# axis=1 означает соединение по горизонтали (колонки к колонкам)
combined_df_С = pd.concat([С_ZG, С_OG], axis=1)

# Если появились дубликаты колонок, их можно удалить так:
combined_df_С = combined_df_С.loc[:, ~combined_df_С.columns.duplicated()]
combined_df_С
combined_df_С = combined_df_С.drop("Имя",axis=1)

pearson_corr_С = combined_df_С.corr(method='pearson')
print("Корреляция Пирсона:")
print(pearson_corr_С)
plt.figure(figsize=(20, 20))

# Создаем тепловую карту
sns.heatmap(pearson_corr_С, annot=True, cmap='coolwarm', vmin=-1, vmax=1, fmt=".2f")
plt.title('Корреляционная матрица признаков датасета')
plt.tight_layout()
plt.show()
```
