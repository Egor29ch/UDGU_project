```python
df_clean = df_balanced.copy()
y = df_clean["здоров(1)/болен(0)"]
df_clean_WO_Y = df_clean.drop("здоров(1)/болен(0)",axis = 1)
```

```python
import pandas as pd
import numpy as np
new_data = {}
def get_top_pair(df):
    """Находит одну пару с самой высокой корреляцией > 0.7"""
    corr_matrix = df.corr().abs()
    # Оставляем только верхний треугольник, чтобы не дублировать пары
    upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
    
    # Разворачиваем матрицу в список пар
    stacked = upper.unstack().dropna()
    high_corr = stacked[stacked > 0.7].sort_values(ascending=False)
    
    if not high_corr.empty:
        # Возвращаем имена столбцов первой пары и значение корреляции
        return high_corr.index[0], high_corr.iloc[0]
    return None, None

def multiply_and_clean(df):
    working_df = df.copy()
    count = 0
    
    while True:
        # 1. Сначала получаем результат в две переменные
        res = get_top_pair(working_df)
        pair_index, val = res
        
        # 2. Проверяем, не пустой ли результат
        if pair_index is None:
            break
            
        # 3. Только ТЕПЕРЬ распаковываем имена столбцов
        col2, col1 = pair_index
        new_col_name = f"{col1}_x_{col2}"
        
        # print(f"Итерация {count+1}: {col1} + {col2} (corr: {val:.3f})")
        
        working_df[new_col_name] = working_df[col1] * working_df[col2]
        working_df = working_df.drop([col1, col2], axis=1)
        
        count += 1

        if count % 20 == 0:
            working_df = working_df.copy()
        
    print(f"Готово! Операций: {count}. Итоговый размер: {working_df.shape}")
    return working_df

# Запуск процесса
df_interacted_WO_Y = multiply_and_clean(df_clean_WO_Y)
```
