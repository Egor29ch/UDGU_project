```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df_all_name = pd.read_excel(r"tables.xlsx",sheet_name="ВСЕ ИМЕНА")
df_biochemistry = pd.read_excel(r"tables.xlsx",sheet_name="БИОХИМИЯ")
df_plasma = pd.read_excel(r"tables.xlsx",sheet_name="ФРАКЦИИ плазма")
df_red_blood_cells = pd.read_excel(r"tables.xlsx",sheet_name="ФРАКЦИИ эритр.")
df_sheet7 = pd.read_excel(r"tables.xlsx",sheet_name="Лист7")

```

```python
df_B_sh = pd.merge(df_biochemistry, df_sheet7, how='outer', on="ФИО", left_on=None, right_on=None, left_index=False, right_index=False)
df_B_sh_all = pd.merge(df_B_sh, df_all_name, how='outer', on="ФИО", left_on=None, right_on=None, left_index=False, right_index=False)
df_B_sh_all_plasma = pd.merge(df_B_sh_all, df_plasma, how='outer', on="ФИО", left_on=None, right_on=None, left_index=False, right_index=False)
df_finaly = pd.merge(df_B_sh_all_plasma, df_red_blood_cells, how='outer', on="ФИО", left_on=None, right_on=None, left_index=False, right_index=False)
```
